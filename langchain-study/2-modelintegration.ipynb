### Models Integration With OpenAI, Google Gemini and GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model=init_chat_model("google_genai:gemini-3.8-flash")

response=model.invoke("Tell me somethign about One Piece Anime")

response.content
```

```python
response.content[0]['text']
```

### STREAMING

```python
model.stream("Tell me something about anime called One Piece.")
```

```python
for chunk in model.stream("Tell me something about anime called One Piece."):
    print(chunk)
```

```python
for chunk in model.stream("Tell me something about anime called One Piece."):
    print(chunk.text,end="|",flush=True)
```

### BATCH

```python
responses=model.batch([
    "Who is Luffy In One Piece?",
    "How old is Goku in dragin ball?",
    "Tell me feats of Ronaldo in 2008",
    "What are the powers of Dr Doom?"
],
config={
    'max_concurrency':2 #limit to 2 parallel calls so 4 questions sent 2 at a time
})

for response in responses:
    print(response)
```

```python
responses = model.batch([
    "Who is Luffy In One Piece?",
    "How old is Goku in dragin ball?",
    "Tell me feats of Ronaldo in 2008",
    "What are the powers of Dr Doom?"
],
config={
    "max_concurrency": 2  # limit to 2 parallel calls
})

for response in responses:
    print(response.text)
    print("-" * 50)
```
