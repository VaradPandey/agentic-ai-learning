```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model=init_chat_model("groq:openai/gpt-oss-20b")

model.invoke("HI")
```

```python
from langchain.messages import SystemMessage,HumanMessage,AIMessage


messages=[
    SystemMessage("You are a professional football coach"),
    HumanMessage("Give detailed analysis of Kylian Mbappe Playstyle")
]

#alternalt
#system_msg=SystemMessage() and then messages=[system_msg]

response=model.invoke(messages)
response.content
```

```python
#use system message for detailed info to llm

system_msg=SystemMessage("""You are an expert in football tactics.
Always provide examples and explain your reasoning.
Be concise but thorough in you explanations.
""")

human_msg=HumanMessage("who is better in attacking and defence lamine yamal vs kyliam mbappe")

messages=[
    system_msg,
    human_msg
]

response=model.invoke(messages)
print(response.content)
```

```python
#manual example ai message eg for conversation history
ai_msg=AIMessage("I would be happy to hep you with that question")

system_msg=SystemMessage("You are a preofessional cook,always give detailed ingredients list, and exact step wise instructions")

human_msg=HumanMessage("tell me how to cook chicken changezi")

messages=[
    system_msg,
    human_msg,
    ai_msg
]

response=model.invoke(messages)
print(response.content)
```

```python
ai_message=AIMessage(
    content=[],
    tool_calls=[{
        "name": "get_weather",
        "args": {
            "location": "Delhi"
        },
        "id": "call_123"
    }]
)

#execute tool an ccreate result message
weather_result="Sunny, 32 C"

from langchain.messages import ToolMessage

tool_message=ToolMessage(
    content=weather_result,
    tool_call_id="call_123" #must match the call id   
)

#continue coversation
messages=[
    HumanMessage("What's the weather in Delhi right now?"),
    ai_message,
    tool_message
]

response=model.invoke(messages)
print(response.content)
```
